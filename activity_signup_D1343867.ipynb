## 作業說明
本作業使用 Python 的字串、list、tuple、dict、set，
整理桌遊體驗與電影欣賞兩場活動的報名資料，
計算參加人數、人次及應收費用。


```python

# HW09：兩場活動報名整理器

# 固定的兩個活動名稱，以 tuple 保存
activity_names = ("桌遊體驗", "電影欣賞")

# 原始報名文字
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " AMY, ben, CORA, amy "

# 每個活動的費用
unit_fees = {
    "桌遊體驗": 120,
    "電影欣賞": 80
}

# 顯示原始資料
print("活動名稱：", activity_names)
print("桌遊原始資料：", board_raw)
print("電影原始資料：", movie_raw)
print("活動費用：", unit_fees)
```

```python

# 從 tuple 取得活動名稱
board_name, movie_name = activity_names

# 整理桌遊名單
board_list = board_raw.replace(" ", "").lower().split(",")

# 整理電影名單
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 顯示整理結果
print("桌遊整理串列：", board_list)
print("電影整理串列：", movie_list)
```

```python

# 將兩場名單轉成集合，去除重複姓名
board_set = set(board_list)
movie_set = set(movie_list)

# 共同參加者：兩場都有參加
common_people = board_set & movie_set

# 全部參加者：至少參加一場
all_people = board_set | movie_set

# 只參加桌遊的人
board_only = board_set - movie_set

# 只參加電影的人
movie_only = movie_set - board_set

# 顯示集合運算結果
print("桌遊集合：", board_set)
print("電影集合：", movie_set)
print("共同參加者：", common_people)
print("全部參加者：", all_people)
print("只參加桌遊者：", board_only)
print("只參加電影者：", movie_only)
```

```python

# 兩場各自去重並排序
board_sorted = sorted(board_set)
movie_sorted = sorted(movie_set)

# 各種集合結果排序
common_sorted = sorted(common_people)
board_only_sorted = sorted(board_only)
movie_only_sorted = sorted(movie_only)
all_sorted = sorted(all_people)

# 取得通知名單前兩位
preview = all_sorted[:2]

# 顯示結果
print("桌遊去重名單：", board_sorted)
print("電影去重名單：", movie_sorted)
print("共同參加者：", common_sorted)
print("只參加桌遊者：", board_only_sorted)
print("只參加電影者：", movie_only_sorted)
print("全部參加者：", all_sorted)
print("通知名單前兩位：", preview)
```

```python

# 計算各項人數
board_count = len(board_set)
movie_count = len(movie_set)
common_count = len(common_people)
total_people = len(all_people)

# 計算報名人次
total_registrations = board_count + movie_count

# 從字典取得每人費用
board_fee = unit_fees[board_name]
movie_fee = unit_fees[movie_name]

# 計算各場應收費用
board_total = board_count * board_fee
movie_total = movie_count * movie_fee

# 計算總應收
total_fee = board_total + movie_total

# 建立摘要字典
summary = {
    "桌遊人數": board_count,
    "電影人數": movie_count,
    "共同人數": common_count,
    "總不重複人數": total_people,
    "報名人次": total_registrations,
    "桌遊應收": board_total,
    "電影應收": movie_total,
    "總應收": total_fee
}

print(summary)
```

```python

print("HW09 活動報名摘要")

print(f"活動：{board_name}，每人 {board_fee:.2f} 元")
print(f"活動：{movie_name}，每人 {movie_fee:.2f} 元")

print("桌遊整理串列：", board_list)
print("電影整理串列：", movie_list)

print("桌遊去重名單：", board_sorted)
print("電影去重名單：", movie_sorted)

print(f"桌遊人數：{summary['桌遊人數']} 人")
print(f"電影人數：{summary['電影人數']} 人")

print("共同參加者：", common_sorted)
print(f"共同人數：{summary['共同人數']} 人")

print("只參加桌遊者：", board_only_sorted)
print("只參加電影者：", movie_only_sorted)

print("全部參加者：", all_sorted)

print(f"總不重複人數：{summary['總不重複人數']} 人")
print(f"報名人次：{summary['報名人次']} 人次")

print("通知名單前兩位：", preview)

print(f"桌遊應收：{summary['桌遊應收']:.2f} 元")
print(f"電影應收：{summary['電影應收']:.2f} 元")
print(f"總應收：{summary['總應收']:.2f} 元")
```
